# **Tuples**

A tuple in Python is similar to a list. The difference between the two is that we cannot change the elements of a tuple once it is assigned whereas we can change the elements of a list.

In short, a tuple is an immutable list. A tuple can not be changed in any way once it is created.

####**Characterstics:**

- Ordered
- Unchangeble
- Allows duplicate

##**Creating Tuple**

```python
#empty tuple
t1 = ()
print(t1)

# creating a tuple with a single item
t2 = (2)
print(type(t2)) # So you can see a tuple with only one item is not treated as tuple.


# So below is the way to make a simple item tuple
t3 = (1,)
print(type(t3))


#Homogenous Tuple
t4 = (1,2,3,4)
print(t4)


#Hetrogenous Tuple
t5 = (1,2.5,'h',True)


#2D Tuple: A tuple within a tuple
t = (1,2,3,(4,5))


# Using Type Conversion
t6 = tuple('hello')
print(t6)
```

## **Difference between Lists and Tuples**

- Syntax
- Mutability (Tuples are immutable)
- Speed (Tuples are faster because they are immutable)
- Memory (Tuple occupy less memory)
- Built in functionality (Tuples have less functionality)
- Error prone (Tuples are easy in debugging)
- Usability (we prefer tuple which are read only and if user make changes or interact more with the app then we use lists)


```python
# Here you can see that tuples are fast although we dont do much calculations but anyway tuples are fast than list.
import time

L = list(range(100000000))
T = tuple(range(100000000))

start = time.time()
for i in L:
  i*5
print('List time',time.time()-start)

start = time.time()
for i in T:
  i*5
print('Tuple time',time.time()-start)
```

```python
# Here you can see tuple occupy less memory
import sys

L = list(range(1000))
T = tuple(range(1000))

print('List size',sys.getsizeof(L))
print('Tuple size',sys.getsizeof(T))
```

```python
# Here you can see list are error prone
a = [1,2,3]
b = a

a.append(4)
print(a)
print(b)
```

```python
# here you can see less chances of mistake in tuple because of immutable
a = (1,2,3)
b = a

a = a + (4,) # here we are creating a completely new tuple with combination of already available tuple and 4.
print(a)
print(b)
```

# **Sets**

A set is an unordered collection of items. Every set element is unique (no duplicates) and must be immutable (cannot be changed).

However, a set itself is mutable. We can add or remove items from it.

Sets can also be used to perform mathematical set operations like union, intersection, symmetric difference, etc.

**Characterstics:**
- Unordered
- Mutable
- No Duplicates
- Can't contain mutable data types

##**Creating Sets**

```python
# empty set
s = {}
print(s)
print(type(s)) # look this is a diectionary

#So the correct way is:
s = set()
print(s)
print(type(s))

#1D & 2D sets
s1 = {1,2,3}
print(s1)

#s2 = {1,2,3,{4,5}} # this will throw error because set cant contain mutable items
s2 = {1,2,3,(4,5)}
print(s2)

#Homogenous and Hetrogenous Sets
s3 = {1,2,3,4} # homogenous
print(s3)

s4 = {1,2.5,True} # hetrogenous
print(s4) # As python treat True as 1 so no duplicates thats why it not printed.

#using Type conversion
s5 = set([1,2,3])
print(s5)

#Set cant have mutable items
s6 = {1,2,3,[4,5]}
print(s6)
```

```python
# Sets are Unordered

s1 = {1,2,3}
s2 = {2,3,1}

print(s1 == s2)
```

##**Accessing Items**

We cant access items from a set. Once an item entered a set there is no way to see it.

```python
s1 = {1,2,3}
print(s1[0]) # This will throw error because sets are unordered dont follow indexing

# You can't do slicing and indexing in sets.
```

##**Editing Items**

```python
s1 ={1,2,3}
s1[1] = 5
# Cant perform in sets because they dont follow indexing because they are unordered.d
```

##**Adding Items**

```python
s1 = {1,2,3}

# Add (anywhere in the set)
s1.add(7)
print(s1)

#Update
#You can add multiple items through update
s1.update([8,9])
print(s1)
```

##**Deleting Items**

```python
# del (delete the complete set)
s = {1,2,3,4,5}
print(s)

s = {1,2,3,4,5}


# del s[0] #cant do any particuar item

# discard
# Here you tell which item you want to delete.
s.discard(5)
print(s)

# If you put any value that is not in the set to discard then nothing happen

# remove
# Work same as Discard
s.remove(4)
print(s)

# If you want to remove any item that is not present then it will throw error.
# And that is the only difference between discard and remove


# pop
# Randomly delete any item
s.pop()
print(s)

# clear
# Empry the whole set
s.clear()
print(s)
```

## **Set Operations**

```python
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

# Union(|)
# Every item of each set printed once without duplication
print(s1 | s2)

# Intersection(&)
# Common items of the set are printed
print(s1 & s2)

# Difference(-)
# Item of s1 that are not in s2
print(s1 - s2)

#item of s2 that are not present in sq
print(s2 - s1)

# Symmetric Difference(^)
# Every item is printed except common item
print(s1 ^ s2)

# Membership Test
print(1 not in s1)

# Iteration
for i in s1:
  print(i)
```

##**Set Functions**

```python
# len/sum/min/max/sorted
s = {3,1,4,5,2,7}

len(s)

sum(s)

min(s)

max(s)

sorted(s,reverse=True)
```

```python
# union/update
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

# s1 | s2 You can also get union of two sets using function
s1.union(s1)


s1.update(s2) # The union operation on set s1 and s2 is performed and result go into s1
print(s1)
print(s2)
```

```python
# intersection/intersection_update
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

s1.intersection(s2)

s1.intersection_update(s2)
print(s1)
print(s2)
```

```python
# difference/difference_update
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

s1.difference(s2)

s1.difference_update(s2)
print(s1)
print(s2)
```

```python
# symmetric_difference/symmetric_difference_update
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

s1.symmetric_difference(s2)

s1.symmetric_difference_update(s2)
print(s1)
print(s2)
```

```python
# isdisjoint/issubset/issuperset
s1 = {1,2,3,4}
s2 = {7,8,5,6}

#isdisjoint
# Mathematiclly disjoint sets are those which doesnot have any common item in them.
print(s1.isdisjoint(s2))

#issubset
# if the 2nd set items are present in 1st set.
s1 = {1,2,3,4}
s2 = {2,3}

print(s2.issubset(s1))


# issuperset
# 1st set contain all the items of 2nd items or more
s1 = {1,2,3,4,5}
s2 = {3,4,5}

print(s1.issuperset(s2))
```

```python
# copy
# Creates a shallow copy at different locations in memory
s1 = {1,2,3}
s2 = s1.copy()

print(s1)
print(s2)
```

# **Dictionary**

Dictionary in Python is a collection of keys values, used to store data values like a map, which, unlike other data types which hold only a single value as an element.

**In some languages it is known as map (hashmap) or assosiative arrays.**

dict = { 'name' : 'muneeb' , 'age' : 33 , 'gender' : 'male' }

**Characterstics:**

- Mutable
- Indexing has no meaning
- keys can't be duplicated
- keys can't be mutable items

##**Creating Dictionaries**

```python
# empty dictionary
d = {}
print(d)
print(type(d))

# 1D dictionary
d1 = { 'name' : 'nitish' ,'gender' : 'male' }
print(d1)

# with mixed keys (hetrogenous)
d2 = {(1,2,3):1,'hello':'world'}
print(d2)

# 2D dictionary -> JSON
s = {
    'name':'Muneeb',
     'college':'GCUF',
     'sem':1,
     'subjects':{
         'dsa':50,
         'stats':51,
         'ml':50
     }
}
print(s)

# using sequence and dict function
d4 = dict([('name','muneeb'),('age',32),(3,3)])
print(d4)

# duplicate keys
d5 = {'name':'muneeb','name':'hussain'} #this will return the last key value
print(d5)

# mutable items as keys
d6 = {'name':'muneeb',(1,2,3):2}
d6 = {'name':'muneeb',[1,2,3]:2} # this will throw error because we are passing list(mutable) as key.
print(d6)
```
