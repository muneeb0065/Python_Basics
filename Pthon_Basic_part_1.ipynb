##**4. Keywords and Identifiers:**

**Keywords:** Certain words which are reserved by the interpreter or compiler which they used them to understand what to do when that keyword is in the python. There are 32 words which are keywords in python and we dont use them as variable name.

**Idenfifier:** The name that you use in the program like the name of the variable or list are identifier.
- you cant start identifier from digit.
- you can use uper case lowercase but in special cases you can onl use underscore (_).
- identifiers cant be keywords.

```python
# Identifiers
# You can't start with a digit
name1 = 'Muneeb'
print(name1)
# You can use special chars -> _
_ = 'nMuneeb'  # Here you use underscore as identifier
print(_)
# identiers can not be keyword
```

#**6. Type Conversion:**

Type Conversion is slight different from Type Casting
- In other languages where there is type casting which means permanently changes the type of an original variable.
But in python we have type conversion where we must store the changes type in another variable because it doesnot changes the original variable.

```python
# Implicit: Which is automatically done by interpreter
print(2+2.5) # python is intellegent so it automaticlly add integer and float

print(type(2), type(2.5))

# But its doesnot work everytime
print(2 + '3')
```

```python
# Explicit: When implicit fail we have to tell manually

a = '4'
print(a)
print(type(a))
a = int(a)
print(a)
print(type(a))
```
