##**1. Operators in Python:**

Two perform operations on two or more things are operators.

- Arithmetic Operators
- Relational Operators
- Logical Operators
- Bitwise Operators
- Assignment Operators
- Membership Operators

```python
# Arithmetic Operators:

a,b = 4,5

#Addition
print(a+b)

#Subtraction
print(a-b)

#Multiplication
print(2*2)

#Division
print(5/2)

#Integer Division
# When you divide two numbers and if the quotient is in decimal then it convert it into integer.
print(5//2)

#Modulus
#Tells the reminder when we divide two numbers
print(5%2)

#Power of Operator
print(6**2)
```

```python
#Relational Operators:
# When we compare two number.

print(3>5)

print(4<5)

print(4>=4)

print(4<=4)

print(4==4)

print(2!=3)
```

```python
#Logical Operators:

# AND operator
print(1 and 0)
print(True and True)

# OR operator
print(1 or 0)

# NOT operator
print(not True)
```

```python
#Bitwise Operators:
#These are special operators that operate on binary values

#Bitwise and operator
#Applying AND operation on binary values of 2 and 3
print(2 & 3)

#Bitwise OR operator
print(2 | 3)

#Bitwise XOR operator
#Convert in Binary IF both are same then 0 and both are different then 1
print(2 ^ 3)

#Bitwise NOT
print(~3)

#Bitwise Left Operator
# Shift the bit left side , 2 bits
print(4 >> 2)

#Bitwise Right
# Shift the bits on right side
print(3 << 1)

```

```python
#Assignment Operators:
# = is a assignment operator to assign the value to the variable

#Some ways

a = 2

a +=2 # it means a = a +2
print(a)

```

```python
#Membership Operators:
# To check if something exists in something or not
# in , not in are membership operators

print('M' in 'Muneeb')
print(1 in [2,3,4,5])

# Here look this is the beauty of python. You dont need to run a loop on each element to check if our required is present in list or string.
# Similarly there are many other things that we use directly instead of coding them from scratch.
```

## **Program:**
 The current population of a town is 10000. The population of the town is increasing at the rate of 10% per year. You have to write a program to find out the population at the end of each of the last 10 years.

```python
# let say population on 9th year = x
# so x + 0.1(x) = 10000
# 1.1x = 10000
# x = 10000/1.1


curr_pop = 10000

for i in range(10,0,-1): # Start = 10, End = 1(0 will be excluded) , stepsize = decrease by 1
  print(i,curr_pop)
  curr_pop = curr_pop/1.1
```

##**Program:**

-> **Find Sequence Sum:** <br>
1/1! + 1/2! + 3/3! +.....

```python
# Solution

num = int(input('Enter n: '))

result = 0
fact = 1

for i in range(1, num+1):
  fact = fact * i
  result = result + i/fact

print(result)
print('factorial of ', num , ' is ', fact)
```

##**for else block:**

When you attach else to a for loop, it explicitly tells Python: **Only run this block if the loop crossed the finish line naturally. If it stopped because of a break, skip this block entirely.**

```python
#break
# A user will give a lower and upper range and you have to print prime numbers between this range

lower = int(input("Enter Lower Range: "))
upper = int(input("Enter Upper Range: "))

for i in range(lower, upper+1):
  for j in range(2, i): # here i holds the current number being checked
    if i%j == 0:
      break
  else:
    print(i)
```

```python
# Continue Statement
# Whenever continue statement hit in the code it skip or ignore all the next code and start the next iteration of the loop.

# In real application when we dont want to display the results that user select not to display through filter or we want to show those products whose stock is empty.

for i in range(1,10):
  if i ==5:
    continue

  print(i)
```

```python
# Pass statement
# You use pass where you want to write the logic later so that the interpreter dont throw error.

for i in range(1, 10):
  pass
```
