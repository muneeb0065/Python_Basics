##**Strings:**
Strings in Python are a sequence of Unicode characters.

**Difference between Unicode and ASCII:**

ASCII(American Standard Code for Information Interchange) <br>
Unicode(Universal Character Set (or Unicode Consortium standard))

The early computers are 8 bit (2^8) mean you can use 8 bits to store only english characters.

But these are not enough to store other characters like special characters  or deal with other large values.

So scientists develop 16 bit system (2^16)

##**Accessing Substring from Strings:**

```python
# Positive Indexing
# Indexing from left to right

s = 'hello world!'

print(s[0])

```

```python
# Negative Indexing
# Indexing from right to left

s = ' hello world'

print(s[-1])
```

```python
# Slicing
# Extracting more then one characters from a string

s = 'hello world'

print(s[0:5]) # start from 0th index upto 4(last not included)
print(s[0:]) # Start from 0 to end

print(s[0:6:2]) # step size +ve

print(s[6:0:-1]) # negatively moving while step size of 1
```

```python
# Reverse a string

print(s[::-1])
```

```python
# Accessing from negative index

print(s[-6:])
```

## **--> Split/Join**

```python
# split
# break your string word by word and add it in a list

s = 'hi my name is muneeb'
s.split() # by default it breaks on space
```

```python
# Write a program that can check whether a given string is palindrome or not.
# abba
# malayalam


s = input('enter the string ')
flag = True

for i in range(len(s)//2):
  if s[i] != s[- i -1]: # basically we are comparing with the last word
    flag = False
    print('Not a Palindrome ')
    break

if flag:
  print('Palindrome')


```

```python
# Write a program to count the number of words in a string without split()

s = input('enter the string ')
L=[]
temp = ''
for i in s:
  if i != " ":
    temp = temp+i

  else:
    L.append(temp)
    temp = '' # again empty the temp

L.append(temp)
print(len(L))
```

```python
# Write a python program to convert a string to title case without using the title()

s = input('enter the string')

L =[]

for i in s.split():
  L.append(i[0].upper() + i[1:].lower())

print(' '.join(L))
```

```python
# Write a program that can convert an integer to string.

number = int(input('enter the number'))

digits = '0123456789'
result = ''
while number != 0:
  result = digits[number % 10] + result # when we take the modulus of 345, We get 5.
                    # in the above line look carefully first you are geting the number and then adding result.
  number = number//10 # when we integer divide 345 by 10, we get 34

print(result)
print(type(result))
```
