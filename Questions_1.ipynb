### Q5:- Write a program to find the simple interest when the value of principle,rate of interest and time period is provided by the user.


````python
# Q5: Simple Interest
# Formula: SI = (P * R * T) / 100

principal = float(input('Enter principal amount: '))
rate = float(input('Enter rate of interest (% per annum): '))
time = float(input('Enter time period (in years): '))

simple_interest = (principal * rate * time) / 100

print(f'Simple Interest is: {simple_interest}')
````

### Q6:- Write a program that will tell the number of dogs and chicken are there when the user will provide the value of total heads and legs.

For example:
Input:
heads -> 4
legs -> 12
<br>
Output:
dogs -> 2
chicken -> 2





````python
# Q6: Dogs and Chickens
# Let d = dogs (4 legs), c = chickens (2 legs)
# d + c = heads => c = heads - d
# 4*d + 2*c = legs => 4*d + 2*(heads - d) = legs => 2*d + 2*heads = legs => 2*d = legs - 2*heads => d = (legs - 2*heads) / 2

heads = int(input('Enter total heads: '))
legs = int(input('Enter total legs: '))

# Check if a valid integer solution is possible
if legs % 2 == 0 and legs >= 2 * heads and legs <= 4 * heads:
    dogs = int((legs - 2 * heads) / 2)
    chickens = heads - dogs
    print(f'dogs -> {dogs}')
    print(f'chicken -> {chickens}')
else:
    print('No valid solution possible for the given heads and legs.')
````

### Q7:- Write a program to find the sum of squares of first n natural numbers where n will be provided by the user.

````python
# Q7: Sum of squares of first n natural numbers
# Formula: [n * (n + 1) * (2n + 1)] / 6

n = int(input('Enter a positive integer n: '))

if n > 0:
    sum_squares = ((n * (n + 1) * (2 * n + 1)) / 6)
    print(f'The sum of squares of first {n} natural numbers is: {sum_squares}')
else:
    print('Please enter a positive integer.')
````

### Q8:- Given the first 2 terms of an Arithmetic Series.Find the Nth term of the series. Assume all inputs are provided by the user.

````python
# Nth term of Arithmetic Series
# Formula: An = a + (n - 1) * d
# Given terms: t1, t2 => common difference d = t2 - t1, first term a = t1

t1 = float(input('Enter the 1st term of the AP series: '))
t2 = float(input('Enter the 2nd term of the AP series: '))
n = int(input('Enter the value of N: '))

common_diff = t2 - t1
nth_term = t1 + (n - 1) * common_diff

print(f'The {n}th term of the series is: {nth_term}')
````

### Q9:- Given 2 fractions, find the sum of those 2 fractions.Take the numerator and denominator values of the fractions from the user.

````python
# For fractions n1/d1 and n2/d2:
# Sum = (n1 * d2 + n2 * d1) / (d1 * d2)
import math

num1 = int(input('Enter numerator of 1st fraction: '))
den1 = int(input('Enter denominator of 1st fraction: '))
num2 = int(input('Enter numerator of 2nd fraction: '))
den2 = int(input('Enter denominator of 2nd fraction: '))

# Calculate numerator and denominator of the resulting fraction
res_num = (num1 * den2) + (num2 * den1)
res_den = den1 * den2

# Simplify the fraction by dividing with Greatest Common Divisor (GCD)
gcd_val = math.gcd(res_num, res_den)
simplified_num = res_num // gcd_val
simplified_den = res_den // gcd_val

print(f'Sum of fractions is: {simplified_num}/{simplified_den}')
````

### Q10:- Given the height, width and breadth of a milk tank, you have to find out how many glasses of milk can be obtained? Assume all the inputs are provided by the user.



Input:<br>
Dimensions of the milk tank<br>
H = 20cm, L = 20cm, B = 20cm
<br><br>
Dimensions of the glass<br>
h = 3cm, r = 1cm

````python

# Tank Dimensions
tank_H = float(input('Enter height of milk tank (cm): '))
tank_L = float(input('Enter length of milk tank (cm): '))
tank_B = float(input('Enter breadth of milk tank (cm): '))

# Glass Dimensions
glass_h = float(input('Enter height of glass (cm): '))
glass_r = float(input('Enter radius of glass (cm): '))

vol_tank = tank_H * tank_L * tank_B
vol_glass = 3.14 * (glass_r ** 2) * glass_h

# Using floor division to get the number of complete glasses we can fill
num_glasses = int(vol_tank // vol_glass)

print(f'Number of glasses of milk that can be obtained: {num_glasses}')
````
